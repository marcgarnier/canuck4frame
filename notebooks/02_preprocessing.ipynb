# 02 · Preprocessing

Clean the raw corpus, drop stubs/duplicates, detect language per article, and split into
`articles_all.csv`, `articles_en.csv`, `articles_fr.csv`.

Accents are kept; stopwords are **not** removed here (BERTopic handles that at modeling time).

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
from src.config import load_config
from src import preprocess

config = load_config('../config.yaml')
written = preprocess.preprocess(config)
written
```

```python
import pandas as pd
df = pd.read_csv(written['all'], parse_dates=['date'])
print('Articles by language:'); print(df['lang'].value_counts())
print('\nArticles by year:'); print(df['year'].value_counts().sort_index())
df.head()
```

```python
import seaborn as sns, matplotlib.pyplot as plt
ax = df.groupby(['year','source']).size().unstack(fill_value=0).plot.bar(stacked=True, figsize=(10,5))
ax.set_title('Article volume by year and outlet'); ax.set_ylabel('Articles'); plt.tight_layout()
```
