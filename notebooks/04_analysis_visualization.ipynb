# 04 · Analysis & Visualization

Temporal evolution of frames, the FR-vs-EN comparison (with a chi-square test),
and descriptive word clouds. Outputs are written to `data/results/`.

```python
import sys, os
sys.path.insert(0, os.path.abspath('..'))
import pandas as pd
from bertopic import BERTopic
from src.config import load_config
from src import visualization, stats

config = load_config('../config.yaml'); res = config['paths']['results']
model = BERTopic.load(str(res / 'bertopic_model'))
df = pd.read_csv(res / 'articles_with_topics.csv', parse_dates=['date'])
topics = df['topic'].tolist()
```

## Frames over time

```python
proportions = visualization.frames_over_time(model, df, topics, res)
proportions
```

## French vs English framing

```python
share = visualization.compare_fr_en(df, topics, model, res)
share
```

### Is the FR/EN difference statistically significant? (chi-square)

```python
chi = stats.chi_square_fr_en(df, topics, model, res)
print(f"chi2 = {chi.chi2:.2f}, p = {chi.p_value:.4f}, dof = {chi.dof}")
print(f"Cramer's V = {chi.cramers_v:.2f} (effect size), significant = {chi.significant}")
if chi.note: print(chi.note)
```

## Word clouds + interactive keyword evolution

```python
visualization.wordclouds(model, res)
visualization.topics_over_time_interactive(model, df, res)
print('Wrote frame_wordclouds.png and topics_over_time.html to', res)
```
